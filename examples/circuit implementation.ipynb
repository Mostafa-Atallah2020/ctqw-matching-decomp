```python
%load_ext autoreload
%autoreload 2

import sys
sys.path.append(f"./../")

from src.graphs import StaticGraph
from src.misc import multi_crx

from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
from scipy.linalg import expm # type: ignore

from qiskit.quantum_info import Operator
import numpy as np
```

```python
edges = {('00', '01')}
G = StaticGraph(edges)
H = G.get_adj_mat()
G.draw()
```

```python
delta_t = 0.1

exact_op = expm(-1j * H * delta_t)
exact_operator = Operator(exact_op)
```

```python
match_qc = QuantumCircuit(G.n_qubits)
match_qc.x(1)
match_qc.crx(2 * delta_t, 1, 0)
match_qc.x(1)

match_qc.draw('mpl')
```

```python
matching_op = Operator(match_qc)

dist = np.linalg.norm(exact_operator.data - matching_op.data, ord=2)
print(f"Operator norm difference between exact and matching circuit: {dist:.3f}")
```
